## Roman to integer
[Link do problema](https://leetcode.com/problems/roman-to-integer/description/)

```python
# input
s = 'XIX'
```

```python
def roman_to_integer(s: str) -> int:
    r_map = {
        'I': 1, 'V': 5, 'X': 10, 
        'L': 50, 'C': 100, 'D': 500, 
        'M': 1000
    }
    result = 0
    for i in range(len(s) -1):
        c = r_map[s[i]]
        n = r_map[s[i + 1]]
        if c < n:
            result -= c
        else:
            result += c
    result += r_map[s[-1]]
    return result
```

```python
roman_to_integer(s)
```
