## Longest common prefix

[Link do problema](https://leetcode.com/problems/longest-common-prefix/)

```python
words = ['flower', 'flown', 'flow']
```

```python
def common_prefix(words: list[str]) -> str:
    smallest_word = min(words, key=len)
    prefix = ''
    for i in range(len(smallest_word)):
        ref = words[0][i]
        for w in words:
            if w[i] != ref:
                return ''
        prefix += ref
    return prefix
```

```python
common_prefix(words=words)
```
